```python
# ============================================================
# 📚 TEXT DIFFICULTY PREDICTOR
# Text and Speech Analysis / NLP Application
# Google Colab - Single Cell
# ============================================================

!pip -q install gradio nltk textstat

import gradio as gr
import nltk
import re
import textstat

# Download required NLTK data
nltk.download("punkt", quiet=True)

# ============================================================
# TEXT DIFFICULTY PREDICTOR
# ============================================================

def predict_difficulty(text):

    if text is None or text.strip() == "":
        return "Please enter some text.", ""

    try:

        # ----------------------------------------------------
        # BASIC TEXT PROCESSING
        # ----------------------------------------------------

        text = text.strip()

        words = re.findall(r"\b[a-zA-Z]+\b", text)

        sentences = re.split(r"[.!?]+", text)
        sentences = [
            s.strip()
            for s in sentences
            if s.strip()
        ]

        word_count = len(words)
        sentence_count = len(sentences)

        if word_count == 0:
            return "No words detected.", ""

        # ----------------------------------------------------
        # WORD FEATURES
        # ----------------------------------------------------

        total_characters = sum(len(word) for word in words)

        average_word_length = (
            total_characters / word_count
        )

        average_sentence_length = (
            word_count / max(sentence_count, 1)
        )

        # Count long words
        long_words = [
            word for word in words
            if len(word) >= 8
        ]

        long_word_percentage = (
            len(long_words) / word_count
        ) * 100

        # ----------------------------------------------------
        # READABILITY SCORE
        # ----------------------------------------------------

        readability_score = textstat.flesch_reading_ease(text)

        # ----------------------------------------------------
        # DIFFICULTY SCORE
        # ----------------------------------------------------

        difficulty_score = 0

        # Average word length
        if average_word_length > 7:
            difficulty_score += 3
        elif average_word_length > 5:
            difficulty_score += 2
        else:
            difficulty_score += 1

        # Average sentence length
        if average_sentence_length > 20:
            difficulty_score += 3
        elif average_sentence_length > 12:
            difficulty_score += 2
        else:
            difficulty_score += 1

        # Long words
        if long_word_percentage > 30:
            difficulty_score += 3
        elif long_word_percentage > 15:
            difficulty_score += 2
        else:
            difficulty_score += 1

        # Readability
        if readability_score < 30:
            difficulty_score += 3
        elif readability_score < 60:
            difficulty_score += 2
        else:
            difficulty_score += 1

        # ----------------------------------------------------
        # CLASSIFICATION
        # ----------------------------------------------------

        if difficulty_score <= 6:

            level = "🟢 EASY"

            reason = (
                "The text contains mostly simple words "
                "and relatively short sentences."
            )

        elif difficulty_score <= 9:

            level = "🟡 MEDIUM"

            reason = (
                "The text contains a mixture of simple "
                "and moderately complex words and sentences."
            )

        else:

            level = "🔴 DIFFICULT"

            reason = (
                "The text contains longer words, "
                "longer sentences, and more complex vocabulary."
            )

        # ----------------------------------------------------
        # RESULT
        # ----------------------------------------------------

        result = f"""
📊 TEXT DIFFICULTY RESULT

Difficulty Level: {level}

📈 Difficulty Score: {difficulty_score}/12

📝 Word Count: {word_count}

📄 Sentence Count: {sentence_count}

📏 Average Word Length: {average_word_length:.2f}

📐 Average Sentence Length: {average_sentence_length:.2f} words

🔤 Long Words: {len(long_words)}

📊 Long Word Percentage: {long_word_percentage:.2f}%

📖 Readability Score: {readability_score:.2f}

💡 Reason:
{reason}
"""

        return level, result

    except Exception as e:

        return (
            "Error",
            f"Something went wrong: {str(e)}"
        )


# ============================================================
# GRADIO WEB APPLICATION
# ============================================================

with gr.Blocks(
    title="Text Difficulty Predictor"
) as app:

    gr.Markdown(
        """
        # 📚 Text Difficulty Predictor

        ### Predict whether a text is Easy, Medium, or Difficult.

        **📝 Text → 🔍 Text Analysis → 📊 Difficulty Score → 🎯 Prediction**
        """
    )

    gr.Markdown(
        """
        ### 📌 Example

        **Easy Text:**

        *The sun rises in the east. Plants need sunlight and water to grow.*

        **Difficult Text:**

        *Artificial intelligence utilizes sophisticated computational
        methodologies to facilitate automated decision-making processes.*
        """
    )

    # --------------------------------------------------------
    # TEXT INPUT
    # --------------------------------------------------------

    text_input = gr.Textbox(
        label="📝 Enter Your Text",
        placeholder="Type or paste your text here...",
        lines=10
    )

    # --------------------------------------------------------
    # BUTTON
    # --------------------------------------------------------

    predict_button = gr.Button(
        "🔍 Predict Difficulty",
        variant="primary"
    )

    # --------------------------------------------------------
    # DIFFICULTY OUTPUT
    # --------------------------------------------------------

    difficulty_output = gr.Textbox(
        label="🎯 Difficulty Level"
    )

    # --------------------------------------------------------
    # DETAILS OUTPUT
    # --------------------------------------------------------

    details_output = gr.Textbox(
        label="📊 Analysis Details",
        lines=15
    )

    # --------------------------------------------------------
    # BUTTON ACTION
    # --------------------------------------------------------

    predict_button.click(
        fn=predict_difficulty,
        inputs=text_input,
        outputs=[
            difficulty_output,
            details_output
        ]
    )

    # --------------------------------------------------------
    # FEATURES
    # --------------------------------------------------------

    gr.Markdown(
        """
        ---

        ### ✨ Features

        📝 Text input
        🔢 Word counting
        📄 Sentence counting
        📏 Average word-length analysis
        🔤 Long-word detection
        📖 Readability analysis
        📊 Difficulty scoring
        🟢 Easy classification
        🟡 Medium classification
        🔴 Difficult classification

        **Text and Speech Analysis Project**
        """
    )

# ============================================================
# LAUNCH APPLICATION
# ============================================================

app.launch(share=True)
```
